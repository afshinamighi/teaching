# Quantum Teleportation: The Art of Moving Without Moving

## Introduction

In this section, we explore quantum teleportation, a protocol for transferring the state of one qubit to another qubit using entanglement and classical communication.
Quantum teleportation does not physically move a qubit from one place to another. Instead, we start with a qubit in a state $\ket{\psi} = \alpha\ket{0} + \beta\ket{1}$,
and use a three-qubit circuit to reproduce this state on another qubit. In doing so, the original quantum state is lost, so the process does not create a copy of the qubit.

We will develop the protocol step by step, first using ket algebra to understand what happens to the three-qubit system and then implementing the same process as a quantum circuit.

## Entanglement: When Two Qubits Cannot Be Separated

Before we start with the problem state and the solution we need to conver a new aspect of quantum computing. We have already worked with multi-qubit states such as $\ket{00}$,$\ket{01}$,$\ket{10}$,$\ket{11}$.
But what does it really mean to describe two qubits as one quantum system?
Let us investigate this algebraically before introducing a formal definition of entanglement.

And this is exactly where entanglement appears.

### Separable vs. Entangled

A two-qubit state is called separable if we can write it as $\ket{\psi}=\ket{\psi_0}\otimes\ket{\psi_1}$
In this case, each qubit can be assigned its own individual quantum state.

For example, $\frac{1}{2}(\ket{00}+\ket{01}+\ket{10}+\ket{11})$ is separable because $\frac{\ket{0}+\ket{1}}{\sqrt{2}}\otimes\frac{\ket{0}+\ket{1}}{\sqrt{2}}$

*An entangled state is a multi-qubit state that cannot be written as a tensor product of individual qubit states.*
For example, $\frac{1}{\sqrt{2}}(\ket{00}+\ket{11})$ is entangled.
The two qubits must now be described as one combined quantum state. We cannot assign a separate pure state to each qubit that completely describes the joint state.
Interestingly, this state has a special name. It is one of the four Bell states, and is denoted by $\ket{\Phi^+}$.

This shared entanglement will become one of the key resources that makes quantum teleportation possible.

## Protocol

### Problem Statement

Suppose Alice has a qubit in an arbitrary state $\ket{\psi} = \alpha\ket{0} + \beta\ket{1}$, where, for now, $\alpha$ and $\beta$ are real numbers satisfying $\alpha^2 + \beta^2 = 1$.

Alice wants Bob to have a qubit in exactly the same state $\ket{\psi}$.
There is one important restriction: Alice cannot send her qubit to Bob.

How can Alice transfer the state $\alpha\ket{0} + \beta\ket{1}$ to Bob’s qubit without physically sending her own qubit?

There are two additional challenges:

* Alice does not need to know the values of $\alpha$ and $\beta$.
* She cannot simply measure her qubit and tell Bob what she observed. A measurement gives only a classical result, $0$ or $1$, and does not reveal the complete quantum state.

This is the problem that quantum teleportation solves.

----------------

### The Setup

To solve the problem, we need three qubits:

* $\ket{q_0}$ — Alice’s qubit containing the state she wants to teleport: $\ket{\psi} = \alpha\ket{0} + \beta\ket{1}$.

* $\ket{q_1}$ — a second qubit belonging to Alice, initially in state $\ket{0}$.
* $\ket{q_2}$ — Bob’s qubit, also initially in state $\ket{0}$.

Therefore, the initial three-qubit state is $\ket{\Psi_0}=(\alpha\ket{0}+\beta\ket{1})\otimes\ket{0}\otimes\ket{0}$.

The important additional resource is that Alice and Bob will share an entangled pair. Qubits $\ket{q_1}$ and $\ket{q_2}$ will be entangled, while Alice’s original qubit $\ket{q_0}$ remains the state that we want to teleport.

We can therefore think of the setup as: $\ket{q_0}$ the state to teleport, $\ket{q_1}$ Alice, $\ket{q_2}$ Bob.
Alice controls $q_0$ and $q_1$, while Bob controls $q_2$.
Our goal is to perform a sequence of operations such that, at the end of the protocol, Bob’s qubit $q_2$ is in the state

$$
\boxed{\alpha\ket{0}+\beta\ket{1}}
$$

without Alice ever sending $q_0$ physically to Bob.


### Building the Solution Step by Step

We start with three qubits: $\ket{\Psi_0}=\ket{q_0}\ket{q_1}\ket{q_2}=(\alpha\ket{0}+\beta\ket{1})\otimes\ket{0}\otimes\ket{0}$.
Our goal is to make $q_2$ end up in the state $\ket{\psi}=\alpha\ket{0}+\beta\ket{1}$.
But, Alice is not allowed to physically send $q_0$ to Bob (the qubit does not travel; the quantum state is transferred.)
Let us build the solution gradually.

#### Step 1 — How can Alice and Bob create a quantum connection?

Before thinking about the state $\ket{\psi}$, consider the following question:
*What resource could Alice and Bob prepare in advance that connects their two qubits in a genuinely quantum way?*
We already know one important mechanism for creating such a connection: **entanglement**.

Alice and Bob therefore prepare $q_1$ and $q_2$ as an entangled pair.
Started initially from $\ket{q_1q_2}=\ket{00}$, Alice first applies $H$ to $q_1$.
Then, she applies CNOT with $q_1$ as control and $q_2$ as target: 

$$\text{CNOT}(\HI\ \ket{0}\ket{0})=\text{CNOT}(ket{+}\ket{0})=\frac{1}{\sqrt{2}}(\ket{00}+\ket{11})$$

The result seems familiar to use, we now have the Bell state:

$$
\ket{\Phi^+}
=
\frac{1}{\sqrt{2}}
(\ket{00}+\ket{11}).
$$

Alice keeps $q_1$, while Bob can take $q_2$.

The complete three-qubit system is therefore:

$$
\ket{\Psi_1}
=
\ket{\psi}\ket{\Psi^+}
=
(\alpha\ket{0}+\beta\ket{1})
\otimes
\ket{\Phi^+}
=
(\alpha\ket{0}+\beta\ket{1})
\otimes
\frac{1}{\sqrt{2}}(\ket{00}+\ket{11})
$$

Expanding it gives:

$$
\ket{\Psi_1}
=
\frac{1}{\sqrt{2}}
\left(
\alpha\ket{000}
+
\alpha\ket{011}
+
\beta\ket{100}
+
\beta\ket{111}
\right).
$$

Notice something important: *we have not teleported anything yet. Bob’s qubit is now part of an entangled pair, but the information contained in $\alpha$ and $\beta$ is still associated with Alice’s original qubit*.
So our next question is:
*How can Alice make her unknown qubit interact with her half of the entangled pair?*

#### Step 2 — How can Alice connect $\ket{\psi}$ to the entangled pair?

Alice now has two qubits:

* $q_0$, containing $\ket{\psi}$,
* $q_1$, her half of the entangled pair.

A natural way to make the state of one qubit influence another is CNOT.

Alice applies: $\text{CNOT}\ket{q_0}\ket{q_1}$ where $q_0$ is the control and $q_1$ is the target.
Before CNOT, we had:

$$
\ket{\Psi_1}
=
\frac{1}{\sqrt{2}}
\left(
\alpha\ket{000}
+
\alpha\ket{011}
+
\beta\ket{100}
+
\beta\ket{111}
\right).
$$

Remember what CNOT does:

* if $q_0=0$, $q_1$ remains unchanged;
* if $q_0=1$, $q_1$ is flipped.

Therefore,

$$
\ket{000}\rightarrow\ket{000},
\quad ,
\ket{011}\rightarrow\ket{011},
\quad ,
\ket{100}\rightarrow\ket{110},
\quad ,
\ket{111}\rightarrow\ket{101}.
$$

The state becomes:

$$
\ket{\Psi_2}
=
\frac{1}{\sqrt{2}}
\left(
\alpha\ket{000}
+
\alpha\ket{011}
+
\beta\ket{110}
+
\beta\ket{101}
\right).
$$

Now the original qubit has interacted with Alice’s half of the entangled pair.
But we are not finished. We need a way to extract useful information from Alice’s two qubits.

#### Step 3 — Why apply a Hadamard gate now?

Suppose Alice simply measures $q_0$ and $q_1$ at this point.
Would those measurements give us everything we need?

Not quite.
The amplitudes $\alpha$ and $\beta$ are still distributed through the state in a form that is not yet convenient for teleportation.
We need one more transformation.
Alice applies $H$ to $q_0$.

Recall $H\ket{0}=\ket{+}$ and $H\ket{1}=\ket{-}$.
Applying this to $\ket{\Psi_2}$ gives:

$$
\ket{\Psi_3}
=
\frac{1}{2}
\left[
\alpha(\ket{000}+\ket{100})
+
\alpha(\ket{011}+\ket{111})
+
\beta(\ket{010}-\ket{110})
+
\beta(\ket{001}-\ket{101})
\right].
$$

This expression looks more complicated than before.
But here comes the key trick.
Instead of grouping the terms according to $\alpha$ and $\beta$, let us group them according to the first two qubits, the two qubits Alice owns.

After rearranging the terms, we obtain:

$$
\ket{\Psi_3}
=
\frac{1}{2}
\Big[
\ket{00}(\alpha\ket{0}+\beta\ket{1})
+
\ket{01}(\alpha\ket{1}+\beta\ket{0})
$$

$$
+
\ket{10}(\alpha\ket{0}-\beta\ket{1})
+
\ket{11}(\alpha\ket{1}-\beta\ket{0})
\Big].
$$

Now stop and look carefully at this expression.
Something remarkable has happened.
The first two qubits belong to Alice. The qubit inside each bracket is Bob’s qubit.

And in the first case we already see:

$$
\alpha\ket{0}+\beta\ket{1}
=
\ket{\psi}.
$$

The state we wanted to teleport has appeared on Bob’s side!
But there are four possibilities.

⸻

Step 4 — What happens if Alice measures her qubits?

Alice now measures $q_0$ and $q_1$.

There are four possible results:

$$
00,\quad01,\quad10,\quad11.
$$

Look again at the state:

$$
\ket{\Psi_3}

\frac{1}{2}
\Big[
\ket{00}(\alpha\ket{0}+\beta\ket{1})
+
\ket{01}(\alpha\ket{1}+\beta\ket{0})
$$

$$
+
\ket{10}(\alpha\ket{0}-\beta\ket{1})
+
\ket{11}(\alpha\ket{1}-\beta\ket{0})
\Big].
$$

Alice’s measurement tells us which version of $\ket{\psi}$ Bob has.

If Alice measures $00$, Bob has

$$
\alpha\ket{0}+\beta\ket{1}

\ket{\psi}.
$$

Excellent. Nothing needs to be done.

But what about the other three outcomes?

This leads to the final puzzle:

Can Bob transform each of these three states back into $\ket{\psi}$ using gates we already know?

⸻

Step 5 — How can Bob recover the original state?

Consider Alice’s result $01$.

Bob has

$$
\alpha\ket{1}+\beta\ket{0}.
$$

Compare this with

$$
\ket{\psi}

\alpha\ket{0}+\beta\ket{1}.
$$

The roles of $\ket{0}$ and $\ket{1}$ have simply been exchanged.

Which gate can undo this?

The $X$ gate:

$$
X(\alpha\ket{1}+\beta\ket{0})

\alpha\ket{0}+\beta\ket{1}.
$$

Now consider result $10$.

Bob has

$$
\alpha\ket{0}-\beta\ket{1}.
$$

Only the sign of the $\ket{1}$ component is wrong.

The $Z$ gate fixes it:

$$
Z(\alpha\ket{0}-\beta\ket{1})

\alpha\ket{0}+\beta\ket{1}.
$$

Finally, for result $11$, Bob has

$$
\alpha\ket{1}-\beta\ket{0}.
$$

Both corrections are needed.

Bob applies $X$ and $Z$ and obtains the original state.

We can summarize the correction process as:

Alice measures	Bob has	Bob applies
$00$	$\alpha\ket{0}+\beta\ket{1}$	$I$
$01$	$\alpha\ket{1}+\beta\ket{0}$	$X$
$10$	$\alpha\ket{0}-\beta\ket{1}$	$Z$
$11$	$\alpha\ket{1}-\beta\ket{0}$	$XZ$

After the appropriate correction, Bob’s qubit is always

$$
\boxed{
\ket{\psi}

\alpha\ket{0}+\beta\ket{1}
}
$$

which is exactly the state Alice originally had.

⸻

Step 6 — But how does Bob know which operation to perform?

There is one final piece of the puzzle.

Bob does not automatically know what Alice measured.

Alice must send her two measurement results to Bob using an ordinary classical communication channel.

For example, she can send the two classical bits

$$
10.
$$

Bob then knows that he must apply $Z$.

This means that quantum teleportation requires two resources:

1. a previously shared entangled pair, and
2. two bits of classical communication from Alice to Bob.

Only after Bob receives those classical bits and performs the appropriate correction does he recover

$$
\ket{\psi}.
$$

Therefore, teleportation does not allow information to travel faster than light.

⸻

What Actually Traveled?

At the beginning, Alice had

$$
\ket{\psi}

\alpha\ket{0}+\beta\ket{1}.
$$

At the end, Bob has

$$
\ket{\psi}

\alpha\ket{0}+\beta\ket{1}.
$$

Yet Alice never sent the original qubit to Bob.

Even more importantly, after Alice performs her measurements, her original qubit no longer contains the state $\ket{\psi}$.

The state has therefore not been copied.

Instead, the combination of

entanglement + quantum operations + measurement + classical communication

allows the quantum state to be reconstructed on Bob’s qubit.

That is the central idea of quantum teleportation.
